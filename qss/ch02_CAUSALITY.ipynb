### 2.5.1. 최저임금과 실업률

```
Card, David, and Alan B. Krueger. 2000. "Minimum Wages and Employment: A Case Study of the Fast-Food Industry in New Jersey and Pennsylvania: Reply." American Economic Review 90 (5): 1397–1420.
DOI: 10.1257/aer.90.5.1397

````python
minwage = pd.read_csv('minwage.csv', dtype={'chain':'category', 'location':'category'})
minwage.info()
````

````python
minwage.shape
````

````python
minwage.describe().round(2)
````

````python
minwage['chain'].value_counts()
````

````python
minwage['location'].value_counts()
````

````python
minwage.select_dtypes(include='category').columns
````

````python
minwageNJ = minwage.loc[minwage.location != 'PA'].copy()
minwagePA = minwage.loc[minwage.location == 'PA'].copy()
````

````python
(minwageNJ['wageBefore'] < 5.05).mean()
````

````python
(minwageNJ['wageAfter'] < 5.05).mean()
````

````python
(minwagePA['wageBefore'] < 5.05).mean()
````

````python
(minwagePA['wageAfter'] < 5.05).mean()
````

````python
minwageNJ['fullPropAfter'] = (
    minwageNJ['fullAfter'] / (minwageNJ['fullAfter'] + minwageNJ.partAfter)
)

minwagePA['fullPropAfter'] = (
    minwagePA.fullAfter / (minwagePA.fullAfter + minwagePA.partAfter)
)

minwageNJ['fullPropAfter'].mean() - minwagePA['fullPropAfter'].mean()
````

### 2.5.2 중첩편향

````python
minwageNJ['chain'].value_counts(sort=False, normalize=True)
````

````python
minwagePA['chain'].value_counts(sort=False, normalize=True)
````

````python
minwageNJ_bk = minwageNJ.loc[minwageNJ.chain == 'burgerking'].copy()
minwagePA_bk = minwagePA.loc[minwagePA.chain == 'burgerking'].copy()

minwageNJ_bk['fullPropAfter'].mean() - minwagePA_bk['fullPropAfter'].mean()
````

````python
minwageNJ_bk_subset = (
    minwageNJ_bk.loc[(minwageNJ_bk.location != 'shoreNJ') & (minwageNJ_bk.location != 'centralNJ')].copy()
)

minwageNJ_bk_subset['fullPropAfter'].mean() - minwagePA_bk['fullPropAfter'].mean()
````

### 2.5.3 전후비교법 및 이중차분법 설계

````python
minwageNJ['fullPropBefore'] = (
    minwageNJ['fullBefore'] / (minwageNJ.fullBefore + minwageNJ.partBefore)
)

NJdiff = minwageNJ.fullPropAfter.mean() - minwageNJ.fullPropBefore.mean()
NJdiff
````

````python
minwagePA['fullPropBefore'] = (
    minwagePA['fullBefore'] / (minwagePA.fullBefore + minwagePA.partBefore)
)

PAdiff = minwagePA.fullPropAfter.mean() - minwagePA.fullPropBefore.mean()

NJdiff - PAdiff
````
