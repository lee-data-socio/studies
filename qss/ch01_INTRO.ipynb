```python
'''
def myfunction(input1, input2, ..., inputN):
    DEFINE `output` USING INPUTS
    return output
'''

def my_summary(x): # function takes one input
    s_out = x.sum()
    l_out = len(x)
    m_out = x.mean()
    out = pd.Series([s_out, l_out, m_out], index = ['sum', 'length', 'mean'])
    return out # end function by calling output
```

```python
z = np.arange(1,11)
my_summary(z)
```
